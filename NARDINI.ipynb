```python
myfeat = [
    'pol-pol','pol-hyd','pol-pos','pol-neg','pol-aro','pol-ala','pol-pro','pol-gly',
    'hyd-hyd','hyd-pos','hyd-neg','hyd-aro','hyd-ala','hyd-pro','hyd-gly',
    'pos-pos','pos-neg','pos-aro','pos-ala','pos-pro','pos-gly',
    'neg-neg','neg-aro','neg-ala','neg-pro','neg-gly',
    'aro-aro','aro-ala','aro-pro','aro-gly','ala-ala','ala-pro','ala-gly',
    'pro-pro','pro-gly','gly-gly',
    'Frac A','Frac C','Frac D','Frac E','Frac F','Frac G','Frac H','Frac I','Frac K','Frac L',
    'Frac M','Frac N','Frac P','Frac Q','Frac R','Frac S','Frac T','Frac V','Frac W','Frac Y',
    'Frac K+R','Frac D+E','Frac Polar','Frac Aliphatic','Frac Aromatic','R/K Ratio','E/D Ratio',
    'Frac Chain Expanding','FCR','NCPR','Hydrophobicity','Disorder Promoting','Iso point','PPII',
    'A Patch','C Patch','D Patch','E Patch','F Patch','G Patch','H Patch','I Patch','K Patch',
    'L Patch','M Patch','N Patch','P Patch','Q Patch','R Patch','S Patch','T Patch','V Patch',
    'Y Patch','RG Frac',
]

our_accessions = set(acc_df["accession"].dropna())
idr_df = speciesdf[speciesdf["Uniprot"].isin(our_accessions)].copy()
idr_df = idr_df.merge(acc_df[["gene","panel","accession"]], left_on="Uniprot", right_on="accession", how="left")

genes_with_idr = set(idr_df["gene"])
no_idr = sorted(set(acc_df["gene"]) - genes_with_idr)
print(f"Genes with at least one IDR: {len(genes_with_idr)} / {len(acc_df)}")
print(f"Genes with no IDR: {len(no_idr)}")
print(no_idr)

core_cols = ["gene","panel","Uniprot","Genes","Cluster Number","Min Inter Clust Dist",
              "Start Pos","End Pos","IDR Length","Full Protein Length","Sequence"]
out_df = idr_df[core_cols + myfeat].copy()
out_df.to_csv("/content/idr_zscores.csv", index=False)
print(f"\nTotal IDRs: {len(out_df)}")
```

```python
cluster_colors = ['#D865A6','#7C2880','#E23A36','#E23A36','#E23A36','#000000','#1F8241','#E23A36',
                   '#E23A36','#000000','#189BD7','#1F8241','#000000','#E23A36','#189BD7','#1F8241',
                   '#1F8241','#189BD7','#000000','#E23A36','#8C8C8C','#7C2880','#D865A6','#000000',
                   '#E23A36','#189BD7','#189BD7','#7C2880','#F58220','#189BD7']

aa_group_colors = {
    'A':'#8c8c8c',
    'C':'#1f8241','H':'#1f8241','N':'#1f8241','Q':'#1f8241','S':'#1f8241','T':'#1f8241',
    'D':'#e23a36','E':'#e23a36',
    'F':'#f58220','W':'#f58220','Y':'#f58220',
    'G':'#d865a6',
    'I':'#231f20','L':'#231f20','M':'#231f20','V':'#231f20',
    'K':'#189bd7','R':'#189bd7',
    'P':'#7c2880',
}

cluster_desc = dict(zip(clusterdef_df['Cluster'].astype(int), clusterdef_df['Description']))

def color_sequence_html(seq):
    if not seq:
        return ""
    spans = []
    curr_color = aa_group_colors.get(seq[0], '#000000')
    curr_chars = seq[0]
    for ch in seq[1:]:
        c = aa_group_colors.get(ch, '#000000')
        if c == curr_color:
            curr_chars += ch
        else:
            spans.append((curr_color, curr_chars))
            curr_color, curr_chars = c, ch
    spans.append((curr_color, curr_chars))
    return "".join(f'<span style="color:{c}">{txt}</span>' for c, txt in spans)

def zscore_color(z, vmax=3.0):
    z = max(-vmax, min(vmax, z))
    if z >= 0:
        t = z / vmax
        return f"rgb(255,{int(255*(1-t))},{int(255*(1-t))})"
    else:
        t = -z / vmax
        return f"rgb({int(255*(1-t))},{int(255*(1-t))},255)"

gene_idrs = {}
for gene, group in idr_df.groupby("gene"):
    records = []
    for _, row in group.sort_values("Start Pos").iterrows():
        zscores = {feat: round(float(row[feat]), 3) for feat in myfeat if pd.notna(row[feat])}
        top_features = sorted(zscores.items(), key=lambda kv: abs(kv[1]), reverse=True)[:8]

        idr_len  = int(row["IDR Length"])
        full_len = int(row["Full Protein Length"])
        ratio    = idr_len / full_len if full_len else 0

        records.append({
            "idr_label":   row["Genes"],
            "cluster":     int(row["Cluster Number"]),
            "cluster_desc": cluster_desc.get(int(row["Cluster Number"]), ""),
            "cluster_color": cluster_colors[int(row["Cluster Number"])],
            "min_inter_clust_dist": round(float(row["Min Inter Clust Dist"]), 3),
            "start": int(row["Start Pos"]),
            "end":   int(row["End Pos"]),
            "idr_length":  idr_len,
            "full_length": full_len,
            "high_ratio":  ratio > 0.5,
            "sequence_html": color_sequence_html(row["Sequence"]),
            "zscores": [{"feature": f, "z": z, "color": zscore_color(z)} for f, z in zscores.items()],
            "top_features": [{"feature": f, "z": z, "color": zscore_color(z)} for f, z in top_features],
        })
    gene_idrs[gene] = records

print(f"Genes with IDR data: {len(gene_idrs)}")
print(f"Total IDR records: {sum(len(v) for v in gene_idrs.values())}")
```

```python
import requests, time

target_genes = acc_df[acc_df["gene"].isin(gene_idrs.keys())]

protein_names = {}
for i, row in enumerate(target_genes.itertuples(), 1):
    try:
        r = requests.get(f"https://rest.uniprot.org/uniprotkb/{row.accession}.json", timeout=15)
        r.raise_for_status()
        protein_names[row.gene] = r.json().get("uniProtkbId","").replace("_HUMAN","")
    except Exception as e:
        protein_names[row.gene] = ""
        print(f"{row.gene}: ERROR {e}")
    time.sleep(0.12)

print(f"Fetched {len([v for v in protein_names.values() if v])}/{len(target_genes)} protein names")
```

```python
import json
from pathlib import Path

html_dir = Path("/content/html/grammars")
html_dir.mkdir(parents=True, exist_ok=True)

template = """<!DOCTYPE html>
<html lang="en"><head>
<meta charset="utf-8">
<meta name="viewport" content="width=device-width,initial-scale=1">
<title>{gene} — NephVar Molecular Grammars</title>
<link rel="preconnect" href="https://fonts.googleapis.com">
<link href="https://fonts.googleapis.com/css2?family=Syne:wght@400;500;600;700&family=DM+Mono:wght@300;400&family=Instrument+Serif:ital@0;1&display=swap" rel="stylesheet">
<style>
:root{{--serif:'Instrument Serif',Georgia,serif;--sans:'Syne',system-ui,sans-serif;
--mono:'DM Mono',ui-monospace,monospace;--accent:#1f6fa8;--ink:#16202b;
--soft:#5d6b78;--faint:#8a96a3;--line:#e6e9ee;--bg:#fbfaf7}}
*{{box-sizing:border-box}}
body{{margin:0;font-family:var(--sans);color:var(--ink);background:var(--bg);-webkit-font-smoothing:antialiased}}
.wrap{{max-width:1240px;margin:0 auto;padding:40px 22px 70px}}
.crumb{{font-family:var(--mono);font-size:11px;letter-spacing:.14em;text-transform:uppercase;color:var(--faint);margin-bottom:16px}}
.crumb a{{color:var(--accent);text-decoration:none}}.crumb a:hover{{text-decoration:underline}}
h1{{font-family:var(--serif);font-weight:400;font-size:40px;margin:0 0 4px}}
h1 em{{font-style:italic;color:var(--accent)}}
.sub{{color:var(--soft);font-size:14px;margin-bottom:28px;line-height:1.6}}
.map-wrap{{background:#fff;border:1px solid var(--line);border-radius:6px;padding:24px;margin-bottom:28px;overflow-x:auto}}
.map-title{{font-family:var(--mono);font-size:10.5px;text-transform:uppercase;letter-spacing:.1em;color:var(--soft);margin-bottom:16px}}
svg.protmap{{display:block;width:100%;min-width:600px;height:90px}}
.idr-card{{background:#fff;border:1px solid var(--line);border-radius:6px;padding:22px;margin-bottom:18px}}
.idr-head{{display:flex;flex-wrap:wrap;align-items:center;gap:10px;margin-bottom:6px}}
.idr-title{{font-family:var(--serif);font-size:22px}}
.clusterbadge{{font-family:var(--mono);font-size:11px;color:#fff;padding:3px 10px;border-radius:99px}}
.idr-meta{{font-family:var(--mono);font-size:11px;color:var(--faint);margin-bottom:14px}}
.cluster-desc{{font-size:13px;color:var(--soft);margin-bottom:14px}}
.caveat{{background:#fff7e6;border:1px solid #f0d9a0;border-radius:4px;padding:10px 14px;
font-size:12.5px;color:#7a5a10;margin-bottom:14px;line-height:1.5}}
.seqbox{{font-family:var(--mono);font-size:11px;line-height:1.9;word-break:break-all;
background:#fafafa;border:1px solid var(--line);border-radius:4px;padding:12px;
max-height:160px;overflow-y:auto;margin-bottom:16px}}
.section-label{{font-family:var(--mono);font-size:10.5px;text-transform:uppercase;letter-spacing:.1em;
color:var(--soft);margin:14px 0 8px}}
.chips{{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:14px}}
.chip{{font-family:var(--mono);font-size:11px;padding:4px 9px;border-radius:4px;border:1px solid var(--line)}}
details{{margin-top:8px}}
summary{{font-family:var(--mono);font-size:11.5px;color:var(--accent);cursor:pointer;padding:6px 0}}
table{{width:100%;border-collapse:collapse;font-size:12.5px;margin-top:10px}}
th{{font-family:var(--mono);text-align:left;padding:8px 10px;background:#f6f5f1;font-weight:400;
font-size:10px;letter-spacing:.08em;text-transform:uppercase;color:var(--soft);cursor:pointer;white-space:nowrap}}
td{{padding:7px 10px;border-top:1px solid var(--line);font-family:var(--mono);font-size:11.5px}}
</style></head>
<body><div class="wrap">
<div class="crumb"><a href="../">NephVar</a> / <a href="./">Molecular Grammars</a> / {gene}</div>
<h1>{gene} <em>{protein_name}</em></h1>
<div class="sub">{panel} panel &middot; {length} aa &middot;
UniProt <a href="https://www.uniprot.org/uniprot/{accession}" target="_blank"
style="color:var(--accent)">{accession}</a> &middot; {n_idrs} IDR{plural}
&middot; <a href="../biophysical/{gene}.html" style="color:var(--accent)">View biophysical profile &rarr;</a></div>

<div class="map-wrap">
  <div class="map-title">IDR map &mdash; colored by GIN molecular grammar cluster</div>
  <svg class="protmap" id="protmap" viewBox="0 0 1000 90"></svg>
</div>

{idr_cards}

</div>
<script>
const idrs   = {idr_json};
const seqlen = {length};

const svg = document.getElementById("protmap");
const ns  = "http://www.w3.org/2000/svg";
const pad = 20, inner = 1000 - pad*2;
const xOf = p => pad + ((p-1)/Math.max(seqlen-1,1)) * inner;

function rect(a){{const e=document.createElementNS(ns,"rect");for(const[k,v]of Object.entries(a))e.setAttribute(k,v);return e;}}
function text(a,t){{const e=document.createElementNS(ns,"text");for(const[k,v]of Object.entries(a))e.setAttribute(k,v);e.textContent=t;return e;}}

svg.appendChild(rect({{x:pad, y:30, width:inner, height:14, fill:"#eef0f2", rx:"2"}}));

idrs.forEach((idr,i) => {{
  const x = xOf(idr.start), w = Math.max(2, xOf(idr.end)-xOf(idr.start));
  const r = rect({{x, y:30, width:w, height:14, fill:idr.cluster_color, opacity:"0.85", rx:"2"}});
  const t = document.createElementNS(ns,"title");
  t.textContent = `IDR${{i+1}} (${{idr.start}}-${{idr.end}}) — Cluster ${{idr.cluster}}: ${{idr.cluster_desc}}`;
  r.appendChild(t);
  svg.appendChild(r);
  svg.appendChild(text({{x:x+w/2, y:26, fill:"#5d6b78", "font-size":"9",
    "font-family":"DM Mono,monospace","text-anchor":"middle"}}, `IDR${{i+1}}`));
}});

svg.appendChild(text({{x:pad, y:62, fill:"#8a96a3", "font-size":"9","font-family":"DM Mono,monospace"}}, "1"));
svg.appendChild(text({{x:pad+inner, y:62, fill:"#8a96a3", "font-size":"9",
  "font-family":"DM Mono,monospace","text-anchor":"end"}}, String(seqlen));

document.querySelectorAll("table[data-sortable]").forEach(table => {{
  let asc = true;
  table.querySelectorAll("th[data-k]").forEach(th => {{
    th.addEventListener("click", () => {{
      const k = th.dataset.k;
      const tbody = table.querySelector("tbody");
      const rows = Array.from(tbody.querySelectorAll("tr"));
      rows.sort((a,b) => {{
        const av = a.dataset[k], bv = b.dataset[k];
        const an = parseFloat(av), bn = parseFloat(bv);
        const cmp = !isNaN(an) && !isNaN(bn) ? an-bn : av.localeCompare(bv);
        return asc ? cmp : -cmp;
      }});
      rows.forEach(r => tbody.appendChild(r));
      asc = !asc;
    }});
  }});
}});
</script>
</body></html>"""

idr_card_template = """
<div class="idr-card">
  <div class="idr-head">
    <span class="idr-title">IDR {idr_num}</span>
    <span class="clusterbadge" style="background:{cluster_color}">Cluster {cluster}</span>
  </div>
  <div class="idr-meta">Residues {start}&ndash;{end} &middot; {idr_length} aa
    ({pct_of_protein}% of protein) &middot; Min inter-cluster distance: {min_dist}</div>
  <div class="cluster-desc">{cluster_desc}</div>
  {caveat}
  <div class="section-label">Sequence</div>
  <div class="seqbox">{sequence_html}</div>
  <div class="section-label">Top exceptional features (|z-score| rank)</div>
  <div class="chips">{top_chips}</div>
  <details>
    <summary>Show all 90 sequence-feature z-scores</summary>
    <table data-sortable>
      <thead><tr><th data-k="feature">Feature</th><th data-k="z">Z-score</th></tr></thead>
      <tbody>{zscore_rows}</tbody>
    </table>
  </details>
</div>"""

generated = []
for gene, idr_list in gene_idrs.items():
    acc_row = acc_df[acc_df["gene"] == gene].iloc[0]
    accession = acc_row["accession"]
    panel     = acc_row["panel"]
    length    = idr_list[0]["full_length"]
    protein_name = protein_names.get(gene, "")

    cards_html = ""
    for i, idr in enumerate(idr_list, 1):
        pct = round(100 * idr["idr_length"] / idr["full_length"], 1)
        caveat = ""
        if idr["high_ratio"]:
            caveat = ('<div class="caveat">Note: this region spans the majority of the protein '
                      'sequence, which may indicate a structured domain (e.g. a collagen triple '
                      'helix or repetitive fibrous domain) that AlphaFold2 monomer predictions '
                      'and sequence-based disorder predictors can misclassify as disordered. '
                      'Interpret this grammar assignment with caution.</div>')

        top_chips = "".join(
            f'<span class="chip" style="background:{f["color"]}">{f["feature"]}: {f["z"]:+.2f}</span>'
            for f in idr["top_features"]
        )
        zscore_rows = "".join(
            f'<tr data-feature="{f["feature"]}" data-z="{f["z"]}">'
            f'<td>{f["feature"]}</td>'
            f'<td style="background:{f["color"]}">{f["z"]:+.3f}</td></tr>'
            for f in idr["zscores"]
        )

        cards_html += idr_card_template.format(
            idr_num=i, cluster_color=idr["cluster_color"], cluster=idr["cluster"],
            start=idr["start"], end=idr["end"], idr_length=idr["idr_length"],
            pct_of_protein=pct, min_dist=idr["min_inter_clust_dist"],
            cluster_desc=idr["cluster_desc"], caveat=caveat,
            sequence_html=idr["sequence_html"], top_chips=top_chips, zscore_rows=zscore_rows,
        )

    idr_json_min = [{"start": d["start"], "end": d["end"], "cluster": d["cluster"],
                      "cluster_desc": d["cluster_desc"], "cluster_color": d["cluster_color"]}
                     for d in idr_list]

    html = template.format(
        gene=gene, protein_name=protein_name, panel=panel, accession=accession,
        length=length, n_idrs=len(idr_list), plural="" if len(idr_list)==1 else "s",
        idr_cards=cards_html, idr_json=json.dumps(idr_json_min),
    )

    (html_dir / f"{gene}.html").write_text(html, encoding="utf-8")
    generated.append(gene)

print(f"Generated {len(generated)} gene pages → {html_dir}/")
```
